2. Building on the [code](https://github.com/bxlab/cmdb-quantbio/blob/main/lectures/python_dicts_file_io/livecoding.ipynb) for reading in a single FASTA sequence, adapt it to read in multiple sequences from a single FASTA file, storing each sequence in a dictionary using the sequence name as the key. Wrap this code in a function such that it takes a file name as the only function argument and returns the dictionary of sequences.

- To check if a line represents the start of a new sequence, consider using the string method `.startswith()`
- Don't forget to close your filestream

```python
# passing the file name into the function

def dict_from_fasta(fasta_file):

    #opening the file name in read format
    fs = open((fasta_file), 'r')

    #creating a dictionary to add to
    dictionary = {}

    #iterate each line in the fasta file
    for line in fs:

        #is the line a name of sequence?
        if '>' in line:
            #add name as key
            seq_name = line.strip()[1:]
            dictionary[seq_name] = ""
        else:
            #add the value to the key that was just named previously
            dictionary[seq_name] += line.strip()

    fs.close()

    return dictionary

sequence_dict = dict_from_fasta('sequences.fa')
print(sequence_dict.keys()) ## verification that keys are present in dictionary
print(sequence_dict['CCDS2.2|Hs110|chr1']) ##verification that value is populated in dictionary
```

3. Wrap the code for reading in the codon table into a function, taking a file name in as the argument and returning the dictionary of codon/amin acid pairs.

```python
#create a function for this

def codon_table_dict(codon_file):

    #open file that was passed in
    fs = open(codon_file, 'r')

    #create dictionary with {'codon' : AA}
    dictionary = {}

    for line in fs:
        #print(line)
        codon, aa= line.strip('\n').split('\t')

        dictionary[codon] = aa

    fs.close()

    return dictionary

codon_dict = codon_table_dict('codons.tsv')
print(codon_dict.keys())
```

4. Write a function for translating the CDS sequences into amino acid sequences. This function will need to take in two arguments, the codon table and the DNA sequence. The DNA sequences contain untranslated sequences (UTRs) at the start and end so you will need to step through the sequences to find the first methionine (M). Likewise, you will need to stop when you encounter the first step codon (*) rather than translating through the end of the sequence.

- Using a `while` loop may be useful for this task, but it is not required as `for` loops can also work
- You will need to consider three different parts reading the sequence:
    1. Have you reached the start of the coding sequence
    2. Do you need to record the current codon's amino acid
    3. Have you reached the end of the coding sequence

```python
def dna_to_cds_AAs(codon_table, dna_seq):

    AA_seq = ''


    #making a string that is all AA from codons
    for base_pos in range(0,len(dna_seq),3):
        codon = dna_seq[base_pos:base_pos+3]
        AA_coded = codon_table[codon]
        AA_seq += AA_coded

    #print(AA_seq)

    final_aa_seq = ''

# setting Bool for recording coding AA
    record = False
# for loop to only capture coding 
    for character in AA_seq:
       # print(character, record) ## debug line
        #set record to true if it reaches a Methynine ( start codon )
        if character == 'M':
            record = True
        elif character =="*": # set to false after recording the stop codon
            final_aa_seq += character
            record = False
        if record == True: # if record is true, then add the character to the final string
            final_aa_seq += character
        elif record == False:
            continue

    return(final_aa_seq)

aa_seq = dna_to_cds_AAs(codon_dict, sequence_dict['CCDS2.2|Hs110|chr1'])
print(aa_seq)
```

5. Finally, put it all together, loading in the FASTA sequences and codon table, and translating the into amino acids. Once you have the amino acid sequences, count the number of times each amino acid is used. Finally, convert these counts into percentages and write them to a tab-separated file with the first column being the amino acid letter and the second column being the percent usage.

- To get the percentages, it will helpful to keep a running total of the number of amino acids as you find the counts
- The dictionary method `.setdefault` may be useful for intializing you count dictionary for each new amino acid
- Don't forget to close your filestream

```python
seq_dict = dict_from_fasta('sequences.fa')
codon_dict = codon_table_dict('codons.tsv')
#print(seq_dict.keys())


aa_counter = {} ######

for key in sequence_dict:
    #print(key)
    aa_seq = dna_to_cds_AAs(codon_dict, sequence_dict[key])

    for character in aa_seq:
        #print(character)
        aa_counter.setdefault(character, 0)
        aa_counter[character] += 1

#print(aa_counter)

total_aa = 0
for item in aa_counter.values():
    total_aa += int(item)

#print(total_aa)

for amino_acid in aa_counter:
    aa_counter[amino_acid] = aa_counter[amino_acid]/total_aa*100

print(aa_counter)

#make the new file
fs = open('amino_acid_percents.tsv', 'w')
for each_aa in aa_counter:
    write_line = f'{each_aa}\t{aa_counter[each_aa]}\n'
    fs.write(write_line)

fs.close()

```
