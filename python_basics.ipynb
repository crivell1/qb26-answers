3. Calculate and report the 1st, 2nd, and 3rd quartiles of gene expression

    - Your expression values will need to be in order to determine quartiles
    - The first quartile is the 0.25 * (n + 1)th item, the second quartile is the 0.5 * (n + 1)th item, etc.
    - Because python starts counting at zero, the ith item is at index i - 1
    - Remember that the get a value from the list at specific index, you can only use intergers, not floats
    - Use a `print` statement to report your answer

```python
# Sort your expression values from lowest to highest

print("exp_data before sorting:", exp_data)
exp_data.sort()
print("exp_data after sorting:", exp_data)

# Determine which position in the list corresponds to the first, second, and third quartiles

first_quart_position = 0.25 * (len(exp_data) + 1)

second_quart_position = 0.50 * (len(exp_data) + 1)

third_quart_position = 0.75 * (len(exp_data) + 1)

# Find the value from those positions in your expression list

first_quart_value = exp_data[int(first_quart_position) - 1]
second_quart_value = exp_data[int(second_quart_position) - 1]
third_quart_value = exp_data[int(third_quart_position) - 1]


# Report your results
print("First quartile value:", first_quart_value)
print("Second quartile value:", second_quart_value)
print("Third quartile value:", third_quart_value)
```

5. Calculate and report the percentage of genes that are on the positive strand

```python
# Calculate the percentage of positive strand genes

print("strandedness:", strandedness)
count_positive_strand = strandedness.count('+')
count_negative_strand = strandedness.count('-')
total_strand = len(strandedness)

percent_pos_strand = count_positive_strand / total_strand * 100
percent_neg_strand = count_negative_strand / total_strand * 100


# Report your results

print("Percentage of positive strand genes:", percent_pos_strand)
print("Percentage of negative strand genes:", percent_neg_strand)
```
